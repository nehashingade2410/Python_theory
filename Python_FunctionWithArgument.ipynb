#**Function with Arguments**

```python
# We have total 4 types of arguments
# 1. Positional args
# 2. Keyword args
# 3. Default args
# 4. Variable length args(Arbitrary arguments --args and kwargs)
```

```python
# Positional arguments in Python are values that we pass to a
# function in a specific order. The order in which we pass the
# arguments matters.
```

```python
# 1. Positional args - in this sequence order matters

# declaration
def info(name,age,salary):
  print('Your name is: ',name)
  print('Your age is: ',age)
  print('Your salary is: ',salary)

# calling
info('Sonali',25,60000)
```

```python
info(45000,'suraj',25) # this is the problem with positional args.
```

```python
# Keyword arguments are passed by naming the parameters when
# calling the function. This lets us provide the arguments
# in any order, making the code more readable and flexible.
```

```python
# 2. Keyword args- where sequence order doesn't matter
# So the above problem is solved using keyword args.
# means with the values will associate specific keyword
info(salary =45000,name='suraj',age=25)
```

```python
# Default arguments is a parameter that have a predefined
# value if no value is passed during the function call.
```

```python
# 3. Default args - in declaration we specify default values
# so if user did not supply the values from calling
# then default values will be taken & if user supplied new
# values from calling then supplied values will be taken

def info(name='Guest',age=18,salary = 25000):
  print('Your name is:',name)
  print('Your age is:',age)
  print('Your salary is:',salary)

# user is not passing the values in calling
info()
```

```python
# user is passing the new values
info('amol',22)
```

```python
info(age=25,salary=55000)
```

```python
def info(name,age=18,salary = 25000):
  # Non-default comes first (name)
  print('Your name is:',name)
  print('Your age is:',age)
  print('Your salary is:',salary)

info('Yuvraj')
```

```python
# Rule: in calling we must call positonal args first n then keyword
def info(name='Guest',age=18,salary = 25000):  # Non-default comes first (name)
  print('Your name is:',name)
  print('Your age is:',age)
  print('Your salary is:',salary)

info('Yuvraj',age = 45,salary=80000)  # Positional first and the keyword is COREECT
```

```python
# But firsts KEYWORD and then POSITIONAL is wrong
info(name ='Yuvraj',age = 45,80000)
```

```python
info(age = 22,salary=45000,name='Gulnaz')
```

```python
# Variable-length arguments (Arbitrary arguments)
# this allow us to pass a number of arguments to function
# this is useful when we don't know in advance how many
# arguments we will need to pass.
# there are 2 types of arbitrary arguments:
# 1. * args( Non-keyword Arguments)
# 2. **kwargs(Keyword Arguments)
```

```python
# 4. Variable length args
# 4.1 Var. len. positional args.: *args --> gives tuple
# 4.2 Var. len. Keyword args. : **kwargs --> gives dict
```

```python
# # 4.1 Var. len. positional args.: *args
def sample(*num):
  print(num)

sample(10)
sample(10,20,30)
sample()
sample('A',[1,2,3],{1:100})
```

```python
# 42. Var. len. Keyword args.
def display(**data):
  print(data)
# in kwargs we must need to pass key=value combination
display(name='Dipali',place = 'Pune',salary=70000)

display(a=100,b=300)

display()
```

```python
# Combine *args and **kwrgs
def test(*args,**kwargs) :
  print(args,kwargs)

test(10,20,30,name='Dipali',place='Pune')
test()
```

```python
# Q. args vs kwargs
# *args --is used to pass a variable number of positional
# arguments to a function
# these arguments are stored as a tuple.

# **kwargs is used to pass a variable number of keyword arguments
#  and these arguments are stored as a dictionary.
```

```python
# Factorial using recursion
# 5! = 5*4*3*2*1   --> n * (n-1)*....
def factorial(n):
  if n == 1 or n==0:
    return 1
  else:
    return n * factorial(n-1)
fact = factorial(5)
fact
```

```python
# 8!
def factorial(n):
  if n ==1 or n ==0:
    return 1
  else:
    return n * factorial(n-1)
fact = factorial(8)
fact
```

```python
# Comparison( or Relational) Operators
# compares values.
# It either returns True or False according to the condition.
```

```python
a = 10
b = 30

print(a > b)   # 10 > 30 --False --10 is greater than 30 --false
print(a < b)   # 10 < 30 --True  --10 is less than 30 --true
print(a == b)   # 10 ==30 -- 10 is equal to 30 --false
print(a != b)   # 10 != 30  --- 10 is not equal to 30 --true
print( a >= b)   # 10 >= 30 ---10 is greater than or equal to 30 --false
print (a <= b)  # 10 <=30  -- 10 is less than or equal to 30 --True
```

```python
# Logical Operators
# Logical operators perform Logical AND,logical OR and
# Logical NOT operation
```

```python
# Precedence means the order of priority in which operations are performed.
```

```python
# The precedence of Logical Operators in Python is as follows:
# Logical not
# logical and
# logical or
```

```python
a = True
b = False
print(a and b) # True and False = False
print(a or b)   # True or False = True
print(not a)  # not true = False
print(not b) # not False = true
```

```python
"""0 = 1 bit
1 = 1 bit
101 = 3 bits
1010 = 4 bits
10101010 = 8 bits = 1 byte
1 KB	= 1,024 bytes
1 MB = 1,024 KB
A bit is the smallest basic unit of digital information.
1 bit = one binary digit, which can be either 0 or 1.

Logical operators (and, or, not) work with truth
values and evaluate conditions

Bitwise operators (&, |, ^, etc.) operate on the binary bits of integers
"""
```

```python
"""Bitwise Operators
 Bitwise operators act on bits and perform bit-by-bit operations
 These are used to operate on binary numbers.
 Bitwise Operators in python are as follows:
1. Bitwise NOT(~)
2. Bitwise AND(&)
3. Bitwise OR(|)
4. Bitwise XOR(^)
5. Bitwise right shift(>>)
6. Bitwise left shift(<<)
 """
```

```python
a = 10

b = 4
print(a & b)  # a AND b
print( a | b)  # a OR b
print(~a)  # NOT a
print(a ^ b) # a XOR b
print(a >>2) # a right shift
print(a << 2) # a left shift
```

```python
"""Identity operator
"is" and "is not" are the identity operators in Python.
They are used to check if two values (or variables) are located on
the same part of the memory.
Two variables that are equal does not imply that they are identical.

is --True if the operands are identical
is not --True if the operands are not identical
"""
```

```python
a = 10
b = 20
c = a  # means c gets the value a,which is 10
print(a is not b)
print(a is c)
```

```python
# The is not operator checks whether two variables refer
# to different objects.
# Here, a refers to 10 and b refers to 20, so output = True

# a is c
# The is operator checks whether two variables refer to
# the same object, not merely whether their values are equal.
```

```python
# Membership Operators
# "in" and "not in" are the membership operators that are used to
# test whether a value or variable is in sequence.

# in -- True if value is found in the sequence
# not in -- True if value is not found in the sequence
```

```python
x = 24
y = 18
list = [10,20,40,33,60,18,55]
if(x not in list):
  print('x is not present in given list')
else:
  print('x is present')

if(y in list):
  print('y is present in given list')
else:
  print('y is not present')
```

```python
# Ternary Operator
# Ternary Operator also known as conditional expressions
# are operators that evaluate something based on a condition
# being true or false. It was added to Python in version 2.5
# It simply allows testing a condition in a single line
# replacing the multiline if else, making the code compact.
```

```python
# syntax: [on_true] if [expression] else[on_false]
a,b = 10,20
min = a if a < b else b
print(min)
```

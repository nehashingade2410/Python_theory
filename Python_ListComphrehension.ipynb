#**List Comprehension**

```python
# It is a simple and easy way to create a list quickly in one line
# basically it is a one line solution to generate a new list
# syntax : [operation_on_var for var in sequence]
```

```python
# List comprehension is a concise way to create new lists by
# applying an expression to each item in an existing iterable
# like a list,tuple or range.
# It helps to write clean,readable and efficient code compared
# to traditional loops.
```

```python
k = ['','amol','','prasad','','suraj']
```

```python
# PS: we want all names in upper case
[nm.upper() for nm in k]
```

```python
k = ['samrudhi','amol','milind','prasad','atharv','suraj','sonali']
```

```python
# PS: give me the (Name in title,len(name))
[(nm.title(),len(nm)) for nm in k ]
```

```python
# PS: Create a new list of sqaure of 1,10 even numbers

[n**2 for n in range(2,11,2)]
```

```python
[(i**2) for i in range(2,11,2)]  # range(start,stop,step)
```

```python
list(range(2,11,2))
```

```python
for i in range(2,11,2):
  print(i,':',i**2)
```

```python
# 1. [expression for val in squence]
# 2.[expression for val in squence if <condition>]
[nm**2 for nm in range(1,11) if nm % 2 ==0]  # square of even number
```

```python
[nm for nm in range(1,11) if nm < 5]  # filter the num < 5
```

```python
# PS: select those names whose len > 5
[nm for nm in k if len(nm)> 5]
```

```python
# PS: fetch those names who ends with 'i'
[nm for nm in k if nm.endswith('i')]
```

```python
num = 7
check = '+ve' if num > 0 else '-ve'
check
```

```python
# Dict comprehension
s = 'python is simple programming'
# PS- count the number of  occurance of a char. and give me dict{'p':3,...}
# first remove space
s = s.replace(' ','')
{char:s.count(char) for char in s}
```

```python
s.count('p')
```

```python
s = 'python is simple programming'
# without replace
{char: s.count(char) for char in s if char != ' '}
```

```python
s = 'python is simple programming'
# PS- create a new dict of vowels present in a given
# string with the count
{char: s.count(char) for char in s if char in 'aeiou'}
```

```python
s = 'pythOn is simplE programmIng'
# PS- create a new dict of vowels present in a given string with the count
s = s.lower()
{char: s.count(char) for char in s if char in 'aeiou'}
```

```python
s = 'python is simple prOgramming'
{char:s.count(char)for char in s if char in 'AEIOUaeiou'}
```

```python
f = 'I go to Pune and pune is good city. I like it much'
#PS: give me the dict of words with its count
words =f.lower().replace('.','').split()
d = {word:words.count(word) for word in words}
print(d)
```

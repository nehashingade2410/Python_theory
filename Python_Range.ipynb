```python
#Return an object that produces a sequence of integers
# from start (inclusive)to stop (exclusive) by step.
# range(i, j) produces i, i+1, i+2, ..., j-1.
# start defaults to 0, and stop is omitted!
# range(4) produces 0, 1, 2, 3.
# These are exactly the valid indices for a list of 4 elements.
# When step is given, it specifies the increment (or decrement).
```

```python
# range(start,stop)
list(range(2,21))
```

```python
# range(start,stop,step)
list(range(2,21,2))
```

```python
# Rules for stepping:

# If we want to go in forward direction means low to high
# then step must be +ve
u =list(range(2,10,1))
print(u)

# if we want to go in backward direction means high to low
# then step must be -ve
v = list(range(10,2,-1))
print(v)
```

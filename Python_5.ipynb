```python
import hashlib
text = 'shridhar'
#MD5
md5_hash = hashlib.md5(text.encode())
print(md5_hash.hexdigest())
```

```python
# hashlib is a Python module used for hashing data using
# algorithms such as MD5,SHA -1, SHA-256,etc

# text = 'shridhar' ---string shridhar is stored in the variable text

# text.encode() ---hash functions work with bytes, so .encode() converts
# 'shridhar'   to b'shridhar'

# md5_hash = hashlib.md5(text.encode())
# generates an MD5 hash object for 'shridhar'

# md5_hash.hexadigest()
# hexadigexs()  ---converts the hash into readable hexadecimal string



# hashlib.md5() is used to generate an MD5 hash. encode() converts the
# string into bytes and hexadigest() returns the hash
# as a hexadecimal string
```

```python
# Set contains -- operations like union, intersection,
# difference, symmetric_difference
# these operations are temporary
```

```python
x1 = {100,200,300}
x2 = {300,800,900,788,100}
```

```python
# Union --combing all element ---"|" union operator
x1.union(x2)
```

```python
# intersection --comman elements in both the sets --"&"
x1.intersection(x2)
```

```python
#difference() : uncommon element/ distinct elements from set 1
# or first set ---"-"
# distinct elements from set 1
x2.difference(x1)
```

```python
x1.difference(x2)
```

```python
x2
```

```python
# if we want disctinct elements from both sets
x1.symmetric_difference(x2)
```

#**Dictionary**

```python
# A dictionary in Python is a collection of key-value pairs
# It allows you to store and retrieve values based on unique keys
# They are useful for tasks like data mapping and
# storing structured information
# key must be unique and python provides built-in methods to
# work with it
```

```python
# syntax : {}
d ={}
type(d)
```

```python
d1 = dict()
d1
```

```python
# it is Mutable
# we can add element in dict but dict follows a pattern
# of key:value
d.update({'name':'Neha','age':27,'salary':80000})
d
```

```python
# It is not only used to add key value pair
# but if we have/know the key then we can update its value
d.update({'salary':100000})
d
```

```python
# like indexing we can use key as an index to replace a value to add a new key: value pair
```

```python
d3 = [{'name_1':'vidya','age':22},{'name_2':'sneha','age':23}]
d3
```

```python
d3[0]['age']=30
d3
```

```python
s3 = {'employee':[{'name_1':'Vidya','age':30},{'name':'Sneha','age':32}]}
type(s3)
```

```python
s3
```

```python
# Duplicate keys are not allowed
{1:100,1:'ABC',1:500} # it takes last recent key:value pair
```

```python
# It has hash table hence it doesn't allow duplicate keys
```

```python
# duplicate values are allowed
k = {100:1,'ABC':1,600:1}
k
```

```python
k['ABC']='xyz' # if key is already present then this will replace the value
k
```

```python
k['name']='AAAA' # if key and value is new then it will add
k
```

```python
{(1,2,3):['Ashvini','Sneha','Vidya']}  # wrong way to add key
```

```python
# dict is mutable in nature
d = {1:100,2:200,3:300}
d
```

```python
# len()--to find how many items present
len(d)
```

```python
# setdefault()--add a new key with default value None
d.setdefault('Neha')
d
```

```python
d.setdefault(8)
d
```

```python
# along with new key if we provide a value then it take
# provided one

d.setdefault('Shri',123)
d
```

```python
# if key is already present, then it just return its value
d.setdefault(1)
d
```

```python
d['Neha']='Pune' # updating an existing value
d
```

```python
d
```

```python
del d[3] # del[] removes an item using its key
d
```

```python
# pop() removes the item with given key and returns its value

d = {1:100,2:200,3:300}
u = d.pop(2)
print(u)
print(d)
```

```python
# popitem() : popitem() removes and returns the last inserted key-value pair
d = {1:100,2:200,3:300}
d.popitem()
```

```python
d
```

```python
# clear() removes all items from the dictionary
d.clear()
d
```

```python
d = {"a": 1, "b": 2}
for key in d:   # returns all keys
  print(key)
```

```python
for values in d.values():  # returns all values
  print(values)
```

```python
for key,value in d.items(): # return all key-value as tuples
  print(key,value)
```

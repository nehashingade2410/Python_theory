#**Iterative statement**

```python
# Loops are used to execute a block of code repeatedly until a condition
# is met or all items in a sequence are processed.
# The main types are For loops(iterating over sequence) and
# while loops (executing code based on a condition)
```

```python
# For Loop--is used to iterate over a sequence such as list,tuple,string
# or range. It executes a block of code once for each item in the
# sequence.
n = 4
for x in range(0,n):
  print(x)
```

```python
#  = 4 stores the ending value for the loop.
# range(0, n) generates numbers from 0 to 3.
# for x in range(0, n): iterates through each number in the range.
```

```python
# Iterating by Index of Sequences
# A for loop can also iterate through sequence elements using their index
# values with the help of range() and len()

a = [1,2,3,4,5]
for i in range(len(a)):
  print(a[i])
```

```python
# len(a) returns the total number of elements in the list.
# range(len(a)) generates index values from 0 to 2.
#a[i] accesses list elements using their index.
```

```python
# While loop
# it repeatedly executes a block of code as long as the given condition
# remains true. when the condition become false, the line immediately after
# the loop in the program is executed.
u = 0
while (u < 3):
  u = u + 1
  print('Neha')
```

```python
# u = 0 initializes the counter variable.
# while(u < 3): runs the loop while the condition is true
# u = u + 1  increases the counter value by 1
```

```python
# Infinite While loop--it runs continously because its condition
# always remains true.
```

```python
# Nested loops -- is a loop inside another loop. The inner loop executes
# completely for every iteration of the outer loop

for i in range(1,6):
   for j in range(i):
       print(i, end= " ")
   print()
```

```python
# 1. for loop - takes all elements till the end
# 2. while loop - is conditional loop, it executes
# infinite so user needs to place a logic
# which will allow user to exit the loop

# The main objective of the loops:
# to iterate over the collection/iterable

# for loop-
# for var in sequence:
#         body
```

```python
# Transfer statements in python alter the normal flow of execution by
# transferring control to another part of the program.
# 3 main transfer statements in Python are break, continue and pass.
```

```python
"""
Transfer statement
1. break- used to exit the current loop
2. continue - used to skip current condition and continue the loop
3. pass- used to create empty block
"""
```

```python
for i in range(7):
  if i == 4:
    break   # exit
  else:
    print(i)

print('Outside the for loop')
```

```python
cart = [1000,699,799,500,1500,400,2000,999, 500]

# stop if purchase is <=500
for items in cart:
  if items <= 500:
    break
  else:
    print('Process the item-Rs.',items)
```

```python
# stop if purchase is <=500
for items in cart:
  if items <= 500:
    continue
  else:
    print('Process the item-Rs.',items)
```

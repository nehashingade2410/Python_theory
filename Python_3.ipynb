```python
# Q . append vs extend?

# append() --adds its argument as a single element at the end
#           of the list
# extend() --iterates over its argument, adding each individual
#           item to the end of the list

# append() --Adds the object as-is to the end of the list
# extend() --Unpacks the object and adds its individual items

# append()--argument type--any data type(integer,string,list,etc)
# extend() --argument type -Must be an iterable(list,tuple,set,string,etc)

# append()--increase the list length by exactly 1
# extend()--increase the list length by k(number of items in the iterable)

# append()--can create nested list
# extend()--keeps the list flat
```

```python
# Q. Shallow copy vs Deep copy

# Shallow copy --copies the outer object
# Deep copy - copies the outer and nested objects

# shallow copy() creates a new object, but the nested objects inside it
# are still shared with the original object

# deepcopy() creates a completely independent copy,
# including all nested objects. so changes made to the deep
# copy do not affect the original object

# shallow copies copies the outer object but shares nested objects
# whereas deep copy creates an independent copy of both the
# outer and nested objects

# shallow = nested objects shared
# deep =completely independent

# copy ---uses--copy.copy()
list1 =[1,2,3,4]
list2 =list1.copy()
# deepcopy --uses---import copy--copy.deepcopy()
import copy
list1 =[1,2,3,4]
list2 =copy.deepcopy(list1)


# shallow --changes to nested objects can affect original
# deep --changes to nested don't affect original

# we use deep copy when we want to modify a copied object
# independently without affecting the original object,
# especially when the object contains nested mutable objects
```

```python
# the main difference between shallow copy and deep copy is how
# they handle nested objects.
# A shallow copy creates a new outer object but keeps
# reference to the nested objects
# therefore modifiying a nested mutable object can affect the original
# A deep copy recursively copies the nested objects as well
# making the copied object completely independent of the original
```

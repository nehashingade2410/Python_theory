```python
# Q. sort(list method) vs sorted(function)
# Both are used to sort elements, but they work differently

# sort()--it is a list method
# sorted() --it is a built-in function

# sort() --works only with list
# sorted() --works with list,tuple,string,set,dictionaries

# sort()--changes the original list
# sorted() --does not change the original

# sort() --returns None
# sorted() --returns a new sorted list
```

```python
list = [1,2,3,4,5]
list.sort() # returns none
```

```python
sorted(list) # returns a new sorted list
```

```python
# Why do we use functions?
# We use functions to organize and reuse code.
# A function allows us to write a block of code once
# and execute it whenever required.
```

```python
# What is the difference between a function definition and function call?

# Function Definition: is where we create/ write the function
# and specify what the function should do
# We use def keyword
def add(a,b):
  return a+b
# here we are defining the function name add

# Function Call: is where we execute/ run the function
result = add(10,20)
print(result)

# Function definition is used to create a function and specify
# its behavior, whereas a function call is used to execute
# that function
```

```python
# What is the difference between parameters and arguments?

# the main difference is where they are used
# 1. parameters : are variables written inside the function
# definition. They receive values when the function is called
def add(a,b):
  return a + b
# here a, b are parameters

# 2. Arguments: are the actual values that we pass to the function
# when calliong it

add(10,20)
# here 10,20 are arguments

# Parameter--used in function definition
#          --Acts as a placeholder
#         --Example: a,b
# Argument ---Used in function Call
#         ---Actual value
#         --Example: 10,20
```

```python
# What is the return statement?
# The return statement is used inside a function to send a
# result/ value back to the place where the function was called

def add(a,b):
  return a + b
result = add(10,20)
print(result)
# return a + b ==sends 30 back
# result ==stores the returned value
# print(result)  ==display 30
```

```python
# return send a value back from a function
# returns immediately end the function
# a function can return multiple values
# If there in no return, Python returns None
# The return statement is used to return a value from a function
# to the caller. It also terminates the execution of the function
```

```python
# What happens if a function has no return statement?
# If a function does not have a return statement,
# Python automatically returns None

def greet():
    print("Hello")
result = greet()
print(result)

# function executes, But it doesn't return any value.
# Therefore, Python returns None automatically.
```

```python
# If a function does not have a return statement, Python
# automatically returns None. The function can still perform
# operation such as printing or modifying data, but it won't
# return a value to the caller.
```

```python
# What is the difference between print() and return?
# print() --displays a value on the screen
# return --sends a value back from the function

# print() --mainly used to show information
# return --used to get a result from a function

# print() --does not stop a function
# return --stop function execution

# print() --cannot normally be stored as the function's result
# return --returned value can be stored in a variable

# print() --returns None
# return --can return any value
```

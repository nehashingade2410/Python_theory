#**Tuple**

```python
# syntax: ()
# tuples are used to store multiple items in a single variable
# Tuple is a collection which is ordered and unchangeable, allow
# duplicate values
# Tuple has same features as that of the list
# Except one - Tuple is Immutable
```

```python
t1 = ['A','B','C']
t1[0] = 'Apple' # it is possible bcz list is mutable
t1
```

```python
t2 = ('A','B','C') # shows erroe as it is immutable
t2[0] = 'Apple'
t2
```

```python
t2.append(100)
t2
```

```python
# tuple doesn't contain any method who can perform
# the mutable operations
# it contains only 2 methods: count() and index()

# tuple items are indexed, the first item has index [0] 2nd has [1] etc
```

```python
# Unchangeable
# Tuples are unchangeable, meaning that we cannot change
# add or remove items after the tuple has been created
```

```python
# allow duplicates
thistuple = ("apple", "banana", "cherry", "apple", "cherry")
print(thistuple)
```

```python
# Tuple Length
# To determine how many itmes a tuple has, use the len() function
tuple = ("apple", "banana", "cherry", "apple", "cherry")
print(len(tuple))
```

```python
# Q. Which one is better? List or Tuple?

# Neither a list nor a tuple is universally 'better', the
# right choice depends entirely on whether your data needs to change

# Neither List nor Tuple is always better
# We use a List when we need to modify, add, or remove elements
# We use Tuple when the data should reamin unchanged
```

```python
# Which one is faster? List or tuple?

# tuple is faster than list because tuples are immutable,
# while list are mutable
# Tuple generally requires less memory and have slightly
# faster access compared to list
# List is mutable, so it needs extra memory and operations to
# support changes

# Tuple --faster
# List --More flexible
```

```python
# Assignment- Q. List vs Tuple?

# The main difference is mutability, list can be changed
# and tuple cannot

# List --Syntax --[]
# tuple --syntax --()

# List --Mutable --yes
# Tuple --Mutable --No

# List --Speed --Slow
# Tuple --Speed --fast

# List --Memory requirement --More
# Tuple --Memory requiremenyt--Less

# List --Methods --More methods are used
# tuple --Methods --Fewer methods are used

# List --used when --data may change
# Tuple --used when --data should reamin fixed
```

```python
# To create a empty tuple, use round bracket with no content
tuple = ()
print(type(tuple))
```

```python
# We can access tuple items by referring to the index number,
# inside square brackets

tuple = ('apple','banana','cherry')
print(tuple[1])
```

```python
print(tuple[-1])  #  negative index means start from the end
```

```python
# we can specify a range of indexes by specifying where to
# start and where to end the range
# When specifying a range, the return value will be a new
# tuple with specified items

tuple = ('apple','banana','cherry','orange','kiwi','melon','mango')
print(tuple[2:5]) # start at index 2 is included and end at index 5 (not included)
```

```python
# To determine if a specified item is present in a tuple use
# 'in' keyword

tuple = ('apple','banana','cherry')
if 'apple' in tuple:
  print('Yes, apple is present')
```
